In voting ensembel we use multiple different or same models and then we use all the models at the same time and based on their result give the final prediction

It can be used for both classification as well regression

![](Images/voting_intuation.png)

For voting ensembel to work well there are 2 assumptions that we should consider:
1. All the models should be different meaning not prediction same thing for same data the more different the models are the better the result.
2. Minimum accuracy of all the models must be greater then 51%

Based on the probability we can find that the probability of finding the maximum probability will me maximum.
