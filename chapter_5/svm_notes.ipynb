# Support vector machines

The handle outliers very effectively and can work on non linear data as well and can be applied to both regression and classification problems

It works on the same idea of the logistic regression improving its idea

SVM wants to classify the data but by selecting the plane or line that has the most distance from the points 

## Working

![](Images/svm_working.png)

Its working is simple we take a hyperplane or a single line denoted by pie and then we do following steps

1. Draw one line parallel to it until it touches the first value in this case first green cross and denote it pie + .
2. Do the same step but this time for the negative.
3. Calculate the distance between the positive and negative this is called the margin

Do this for all the different hyperplanes that are possible

We will select the hyperplane for which the value of the d or margin is maximum

## Support vectors

## Soft margin svm

Before soft margin svm we will do some changes

- Maximizing a function f(X) is same as minimizing the function 1 / f(x) 

so instead of argmax 2 / || w || we will do argmin of || w || / 2

![](Images/softmargin_equation.png)

Now in the soft margin svm we will add zeta as well and will try to minimize it as welll and this zeta will be calculated for all the points 

if all the points are correctly classified the zeta will be 0

The zeta is basically the distance of wrongly classfied points from its axis positive points from positive axis and negative points from negative axis

![](Images/zeta_distance.png)

The black lines represent the distance between the points zeta

and then we will sum up all these distances and then try to minimize them as well

The svm error is the combination of margin error as well as the classification error as well and we try to minimize both margin error is that we try to increase the margin and the classification error is the wrongly classified example

![](Images/final_zeta.png)

We add c to the zeta as its the hyperparameter if we give the c a greater value the algorithm will focus on minimizing the classification error more then the margin error

## Kernal trick svm

till now what we have discussed the svm was focusing on linear data only but it can also work on the non linear data as well using the kernal trick svm

![](Images/non_linear_data.png)

The above data is the non linear meaning that we can not devide it into two classes using a single line so now we will use the non linear function

In it what we will do is that we  will find a function that will transform the lower dimention data into higher dimension data so it can be transformed this function is called the kernal which transform the data from lower dimension into higher dimension

The transformation of data from lower to higher dimension is called the kernal transformation
