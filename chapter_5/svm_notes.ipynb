# Support vector machines

The handle outliers very effectively and can work on non linear data as well and can be applied to both regression and classification problems

It works on the same idea of the logistic regression improving its idea

SVM wants to classify the data but by selecting the plane or line that has the most distance from the points 

## Working

![](Images/svm_working.png)

Its working is simple we take a hyperplane or a single line denoted by pie and then we do following steps

1. Draw one line parallel to it until it touches the first value in this case first green cross and denote it pie + .
2. Do the same step but this time for the negative.
3. Calculate the distance between the positive and negative this is called the margin

Do this for all the different hyperplanes that are possible

We will select the hyperplane for which the value of the d or margin is maximum

## Support vectors

![](Images/support_vectors.png)

In the first and second step when our line touches a point these points which the line touches are called the support vectors

## Mathmatics of SVM

Before svm one thing we should know is the decesion rule its a rule on the base of which we classify the new points

![](Images/decesion_rule.png)

This is the decesion rule if we have a new point we take its dot product with the w and then add the b to it if its greater then 0 then the point is considerred positive else negative

![](Images/decesion_rule_plot.png)

This is the rule based on the 2d using the equation we can simply see the rule like this like in 2d this rule is also applicable in higher dimensions as well

![](Images/support_vector_decesion_rule.png)

now the decesion rule is done we have to find the values of w and b so we can properly classify the new examples that we have to do with the help of support vector machines

In svm our main goal is to maximize the margin d which is the distance between the 2 hyperplanes positive pie and negative pie in order to do so we have to find first the equations of positive and negative pie and they are

wx + b = 1 (for positive)
wx + b = -1 (for negative)

![](Images/final_equation.png)

These equations are beneficial beacuse if we multiply them by factor less then 1 like 0.5 the distance between them increases meaning the margin increases and if we multiply them by factor greater then 1 then the distance between them decresses meaning the margin decreases so using them we can easily set the margin
